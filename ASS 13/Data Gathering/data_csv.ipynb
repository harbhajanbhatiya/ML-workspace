### Want to change only a specific value of a specific column 
### Converters ---> Help to transforamtion of specific value into a column... It is a function

```python
def station(value):
    if value=="Q":
        return "Queenstown"
    elif value=="S":
        return "Southampton"
    else:
        return value
```

```python
pd.read_csv('titanic.csv',converters={'Embarked':station})
```

### Dataset with missing value or NAN Values if NAN values are represented by ;,:,/,... This can be done by na_values
### Fillna() Or dropna()
