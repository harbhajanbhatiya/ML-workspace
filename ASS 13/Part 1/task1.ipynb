## PART 2 ---> Feature Engineering & Feature Selection

```python
df["Total_Cost"]=df["Tuition_USD"]+df["Rent_USD"]+df["Visa_Fee_USD"]  #to create a new column 'Total_Cost' as the sum of 'Tuition_USD', 'Rent_USD', and 'Visa_Fee_USD'
```

## Feature Selection ---> Mention All Column Names inside the list and that list should be inside the DataFrame

### Seperate I/P & O/P Data Then Feature Scailing

```python
X=df.iloc[:,:8]  #to get all rows and first 7 columns of the dataframe ---> I/P Columns
Y=df["Total_Cost"]  #to get all rows and 'Total_Cost' column of the dataframe ---> O/P Column
```

```python
X
```

```python
Y
```

### Feature Scailing i.e 1 Column value are in ones and 2 Column value are in 2k,3k.... 
### It Help us to keep this data in similar range 
### Two Ways ---> Standaraization (Mean=0 & Std Deviation=1) or Normalization (Mostly Used in Neural Network)
### Don't Standarize your O/P Column because we need to predict it and it should be same amount 

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X=scaler.fit_transform(X)
X
```

### Train & Test Split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
X_train.shape  #80% of data for training
```

```python
X_test.shape  #20% of data for testing
```

#### Mean-absolute-error

```python
# Linear Regression Model
from sklearn.linear_model import LinearRegression
lr=LinearRegression() #model object created
```

```python
lr.fit(X_train,Y_train)  #fitting the model with training data
```

```python
y_predict=lr.predict(X_test)  #predicting the output for test data
```

```python
from sklearn.metrics import mean_absolute_error
mae=mean_absolute_error(Y_test,y_predict)  #calculating mean absolute error
mae #Quite a huge error due to presence of outliers in the data
```

##  Ridge Regression

```python
from sklearn.linear_model import Ridge
ridge=Ridge(alpha=1.0)  #Ridge Regression model object created & alpha is regularization parameter which helps to control the amount of shrinkage applied to the coefficients
```

```python
ridge.fit(X_train,Y_train)  #fitting the Ridge model with training data
```

```python
y_predict_ridge=ridge.predict(X_test)  #predicting the output for test data using Ridge model
```

```python
mae_ridge=mean_absolute_error(Y_test,y_predict_ridge)  #calculating mean absolute error for Ridge model
mae_ridge  #Ridge regression has reduced the error compared to Linear Regression model
```

### Stacking Regressor ---> Ensemble Learning ---> Multiple model of our own Choice

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import StackingRegressor
from sklearn.ensemble import BaggingRegressor
#Creating base models for stacking
```

```python
estimators=[('rf', RandomForestRegressor()),('br',BaggingRegressor())]
```

```python
StackingRegressor(estimators=estimators, final_estimator=DecisionTreeRegressor())
```

```python
stack=StackingRegressor(estimators=estimators, final_estimator=DecisionTreeRegressor())
```

```python
stack.fit(X_train,Y_train)  #fitting the Stacking model with training data
```

```python
y_predict_stack=stack.predict(X_test)
```

```python
mae_stack=mean_absolute_error(Y_test,y_predict_stack) #calculating mean absolute error for Stacking model
mae_stack
```
