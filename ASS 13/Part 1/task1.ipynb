## EDA ---> Univariate Analysis ---> Univariate analysis is the simplest form of quantitative data analysis, focusing on describing, summarizing, and identifying patterns within a single variable at a time. It examines the distribution, central tendency (mean, median, mode), and dispersion (range, variance, standard deviation) of data, such as analyzing only the salaries in a company, without exploring relationships or causes. 
## Bivariate Analysis ---> Bivariate analysis is a statistical method examining the relationship between exactly two variables (X and Y) to determine if they are correlated, associated, or causally related. It determines the strength and direction of these relationships, often identifying one as independent and the other as dependent
