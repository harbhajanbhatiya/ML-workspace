```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

```python
df=pd.read_csv('student_depression_dataset.csv')
df.head()
```

```python
df.info()
```

```python
df.columns
```

### Feature Selection

```python
df=df[['Gender', 'Age', 'City', 'Profession', 'Academic Pressure', 'CGPA', 'Study Satisfaction',
       'Sleep Duration', 'Dietary Habits', 'Work/Study Hours',
       'Financial Stress', 'Family History of Mental Illness', 'Depression']]
```

```python
from sklearn.preprocessing import LabelEncoder 
le = LabelEncoder()
df["Gender"] = le.fit_transform(df["Gender"])
df["City"] = le.fit_transform(df["City"])
df["Profession"] = le.fit_transform(df["Profession"])
df["Dietary Habits"] = le.fit_transform(df["Dietary Habits"])
df["Sleep Duration"]= le.fit_transform(df["Sleep Duration"])
df["Family History of Mental Illness"] = le.fit_transform(df["Family History of Mental Illness"])
```

```python
df.head()
```

```python
df.info()
```

```python
df=df[df["Financial Stress"]!='?'] # Removing rows with missing values in 'Financial Stress' column
df["Financial Stress"]=df["Financial Stress"].astype('float32')
```

```python
df.info()
```

### Sepearting I/P & O/P

```python
X=df[['Gender', 'Age', 'City', 'Profession', 'Academic Pressure', 'CGPA', 'Study Satisfaction',
       'Sleep Duration', 'Dietary Habits', 'Work/Study Hours',
       'Financial Stress', 'Family History of Mental Illness']]
Y=df['Depression']
```

### Feature Scailing

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X=scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train, Y_train)
```

```python
y_pred_lr=lr.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_lr = accuracy_score(Y_test, y_pred_lr)
print(f'Logistic Regression Accuracy: {accuracy_lr:.2f}')
```

### Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(max_depth=5)
dt.fit(X_train, Y_train)
```

```python
y_pred_dt=dt.predict(X_test)
```

```python
accuracy_dt = accuracy_score(Y_test, y_pred_dt)
print(f'Decision Tree Accuracy: {accuracy_dt:.2f}')
```

### Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
rf.fit(X_train, Y_train)
```

```python
y_pred_rf=rf.predict(X_test)
```

```python
accuracy_rf=accuracy_score(Y_test, y_pred_rf)
print(f'Random Forest Accuracy: {accuracy_rf:.2f}')
```

### XGB Classifier, AdaBoost Classifer, CatBoost Classifier

```python
from catboost import CatBoostClassifier
cb = CatBoostClassifier(iterations=100)
cb.fit(X_train, Y_train)
```

```python
y_pred_cb=cb.predict(X_test)
accuracy_cb=accuracy_score(Y_test, y_pred_cb)
print(f'CatBoost Classifier Accuracy: {accuracy_cb:.2f}')
```

```python
from xgboost import XGBClassifier
xgb = XGBClassifier(n_estimators=100)
xgb.fit(X_train, Y_train)
```

```python
y_pred_xgb=xgb.predict(X_test)
accuracy_xgb=accuracy_score(Y_test, y_pred_xgb)
print(f'XGBoost Classifier Accuracy: {accuracy_xgb:.2f}')
```

```python
from sklearn.ensemble import AdaBoostClassifier
adb = AdaBoostClassifier(n_estimators=100)
adb.fit(X_train, Y_train)
```

```python
y_pred_adb=adb.predict(X_test)
accuracy_adb=accuracy_score(Y_test, y_pred_adb)
print(f'AdaBoost Classifier Accuracy: {accuracy_adb:.2f}')
```

### K-NN

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, Y_train)
```

```python
y_pred_knn=knn.predict(X_test)
accuracy_knn=accuracy_score(Y_test, y_pred_knn)
print(f'K-Nearest Neighbors Accuracy: {accuracy_knn:.2f}')
```

### Support Vector Classifier

```python
from sklearn.svm import SVC
svc = SVC(kernel='linear')
svc.fit(X_train, Y_train)
```

```python
y_pred_svc=svc.predict(X_test)
accuracy_svc=accuracy_score(Y_test, y_pred_svc)
print(f'Support Vector Machine Accuracy: {accuracy_svc:.2f}')
```

### Naive Bayes ---> Bernoulli Naive Bayes, Multinomial Naive Bayes

```python
from sklearn.naive_bayes import BernoulliNB
nb = BernoulliNB()
nb.fit(X_train, Y_train)
```

```python
y_pred_nb=nb.predict(X_test)
accuracy_nb=accuracy_score(Y_test, y_pred_nb)
print(f'Naive Bayes Accuracy: {accuracy_nb:.2f}')
```

```python
from sklearn.naive_bayes import MultinomialNB
mnb = MultinomialNB()
mnb.fit(X_train, Y_train) ### In this algo, we can't have negative values in X_train
```

### bagging Classifer

```python
from sklearn.ensemble import BaggingClassifier
bagging = BaggingClassifier(n_estimators=100, random_state=42)
bagging.fit(X_train, Y_train)
    
```

```python
y_pred_bagging=bagging.predict(X_test)
accuracy_bagging=accuracy_score(Y_test, y_pred_bagging)
print(f'Bagging Classifier Accuracy: {accuracy_bagging:.2f}')
```

### Stacking Classifier

```python
from sklearn.ensemble import StackingClassifier
estimators = [                ## Another way to do stacking ---> estimators=[
    ('rf', rf),                                          #       ('rf',RandomForestClassifier(),
    ('bagging', bagging)                                 #       ('bagging',BaggingClassifier())
]
stack=StackingClassifier(estimators=estimators, final_estimator=AdaBoostClassifier())
stack.fit(X_train, Y_train)
```

```python
y_pred_stack=stack.predict(X_test)
accuracy_stack=accuracy_score(Y_test, y_pred_stack)
print(f'Stacking Classifier Accuracy: {accuracy_stack:.2f}')
```

### Hyper Parameter Tuning ---> RandomSearchCV, GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
param_grid = {'n_estimators': [100, 200, 300],'max_depth': [4, 6, 8]}
grid_search = GridSearchCV(estimator=RandomForestClassifier(),param_grid=param_grid,cv=5)
grid_search.fit(X_train, Y_train)
```

```python
grid_search.best_params_ #It will give best parameters found from the grid search
```

```python
grid_search.best_score_ #It will give best score achieved during the grid search
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
param_grid = {'n_estimators': [100, 200, 300],'max_depth': [4, 6, 8]}
randomcv=RandomizedSearchCV = RandomizedSearchCV(estimator=RandomForestClassifier(),param_distributions=param_grid,n_iter=10,cv=5)
randomcv.fit(X_train, Y_train)
```

```python
randomcv.best_params_
```

```python
randomcv.best_score_ 
```
